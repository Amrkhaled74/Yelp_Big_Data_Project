```python
import pyspark.sql.functions as F

def clean_yelp_cities(df, city_col="city"):
    """
    Cleans and standardizes the city column by removing geographic noise, 
    trailing punctuation, state abbreviations, and mapping known typos.
    """
    
    # 1. Define Regex Patterns for Systematic Cleaning
    regex_patterns = {
        # Strip Census noise like "(balance)" or "metropolitan government (balance)"
        "census_balance": r"(?i)\s*(metropolitan government\s*)?\(balance\)",
        
        # Strip state suffixes accidentally appended to the city (e.g., ", FL", ", Arizona")
        "state_bleed": r"(?i),\s*(AL|AK|AZ|AR|CA|CO|CT|DE|FL|GA|HI|ID|IL|IN|IA|KS|KY|LA|ME|MD|MA|MI|MN|MS|MO|MT|NE|NV|NH|NJ|NM|NY|NC|ND|OH|OK|OR|PA|RI|SC|SD|TN|TX|UT|VT|VA|WA|WV|WI|WY|Arizona|California|Florida|Pennsylvania|Tennessee|Missouri)\b",
        
        # Clean parenthetical neighborhood artifacts: "Boise (Meridian)" -> "Boise"
        "parenthetical_neighborhoods": r"\s*\([^)]*\)",
        
        # Clean multi-area slashes: "Wayne/Radnor" -> "Wayne"
        "slash_secondary_areas": r"/.*$",
        
        # Remove trailing punctuation (commas, periods, slashes) and trailing spaces
        "trailing_punctuation": r"[,./\s]+$"
    }

    # 2. Define Explicit Mapping for Typos and Standardization
    # These exact string replacements happen AFTER the regex cleaning
    city_mapping = {
        "Town & County": "Town and Country",
        "Town & Country": "Town and Country",
        "Maryland Height": "Maryland Heights",
        "Mt Laurel Twp": "Mount Laurel",
        "Santa Barbara & Ventura Counties": "Santa Barbara",
        "Land o' Lakes": "Land O' Lakes"
    }

    # 3. Apply Regex Transformations
    cleaned_df = df.withColumn(city_col, F.trim(F.col(city_col)))
    
    # Chain the regex replacements dynamically
    for rule_name, pattern in regex_patterns.items():
        print(rule_name,"        " ,  pattern)
        cleaned_df = cleaned_df.withColumn(
            city_col, 
            F.regexp_replace(F.col(city_col), pattern, "")
        )
        
    # Final trim to clean up any leftover spaces from regex removals
    cleaned_df = cleaned_df.withColumn(city_col, F.trim(F.col(city_col)))

    # 4. Apply Dictionary Mapping for Exact String Matches
    # Using df.replace is much cleaner than chaining dozens of F.when() statements
    cleaned_df = cleaned_df.replace(to_replace=city_mapping, subset=[city_col])

    return cleaned_df

# ==========================================
# Execution Example
# ==========================================
df_business_cleaned = clean_yelp_cities(df_read_business_data, city_col="city")
```

```python
def isValidTime(time):
    
    # Regex to check valid time in 24-hour format. 
    regex = "^([01]?[0-9]|2[0-3]):[0-5][0-9]$"; 

    # Compile the ReGex 
    p = re.compile(regex); 

    # If the time is empty 
    # return false 
    if (time == "") :
        return False;
        
    # Pattern class contains matcher() method
    # to find matching between given time 
    # and regular expression.
    m = re.search(p, time);
    
    # Return True if the time
    # matched the ReGex otherwise False
    if m is None :
        return False
    else :
        return True
```

```python
# This function is used to get subcategory names from the 'categories' column

#@F.udf(returnType=ArrayType(StringType())) # Define DataFrame schema
def get_category_name(text):
    if text is None:
       return "Not Specified",["Not Specified"]  ,["Not Specified"]  
    st = str(text).split(",")
    name=" "
    start_date =[]
    end_date = []
    day=[]
    for i in st:
        stt = str(i).lower().strip()
        if ( "Friday" not in str(text) ) and ( "friday" not in day ) :
            day.append("friday")
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue

        if ("Saturday"  not in str(text) ) and ("saturday"  not in day):
            day.append("saturday")
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue      


        if ( "Sunday"  not in str(text)  ) and  ("sunday"  not in day):
            day.append("sunday") 
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue      

        if ("Monday"  not in str(text) ) and  ("monday"  not in day):
            day.append("monday")
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue      

        if ("Tuesday"  not in str(text) ) and  ("tuesday"  not in day):
            day.append("tuesday")
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue      
      
        if ("Wednesday"  not in str(text) ) and  ("wednesday"  not in day) :
            day.append("wednesday")
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue      

        if ("Thursday"  not in str(text)) and  ("thursday"  not in day):
            day.append("thursday")
            start_date.append("Not specified")
            end_date.append("Not specified")
            continue      


        stt =str(i).strip()
        sttyy=stt.split("'")
        
        for j in sttyy:
            st1 = str(j).split("-")  
            lower_day = j.strip().lower()

            if "friday" in lower_day:
              name = "friday"

            if "tuesday" in lower_day:
              name = "tuesday"           

            if "saturday" in lower_day:
              name = "saturday"                       

            if "sunday"   in lower_day:
              name = "sunday"    

            if "monday"   in lower_day:
                name = "monday"  
           
            if "tuesday"   in lower_day:
               name = "tuesday"

            if "wednesday"   in lower_day:
               name = "wednesday"  

            if "thursday"   in lower_day:
               name = "wednesday"  

            if len(st1)>1:
                count = 0  
                for t in st1:
                    num_1 = t[-2:].strip() 
                    num_2 = t[-1:].strip()
                    num_4 = t[:2].strip()
                    num_3 = t[:1].strip() 
                    time_1 = 0
                    time_2 = 0
                        
                    if num_2.isnumeric():
                        time_1 = num_2
                        
                        if num_1.isnumeric():
                                time_1 = num_1
                    if num_3.isnumeric():
                        time_2 = num_3

                        if num_4.isnumeric():
                                time_2 = num_4  
                    if int(time_2)>=0 and int(time_2) <=11:
                        if int(time_2)== 0: 
                            time_2 = "12"
                        str_1 = time_2 + " : " + time_1 + " AM"
                    
                        if count == 0 :
                              start_date.append(str_1)
                              count = count +1 
                              continue  
                        elif  count == 1:
                              end_date.append(str_1)
                            

                    if int(time_2)>=12 and int(time_2)<=23:

                        if int(time_2) > 12 :
                            time_2 = str(int(time_2) - 12)  
                        str_1 = time_2 + " : " + time_1 + " PM"
                
                        if count == 0 :
                              start_date.append(str_1)
                              count = count +1 
                              continue   
                        elif  count == 1:
                              end_date.append(str_1)
        day.append(name)                          

    return  day,start_date,end_date       

schema = StructType([
    StructField("Day_name", StringType(), True), 
    StructField("shift_start_hour", ArrayType(StringType()), True),
    StructField("shift_end_hour", ArrayType(StringType()), True),  # <-- fixed
])

get_category_name_udf = F.udf(get_category_name, schema)

df_elite = (
    df_hours.withColumn("new_eilte", get_category_name_udf(F.col("hours")))
      .withColumn("Day_name", F.col("new_eilte.Day_name"))
      .withColumn("shift_start_hour", F.col("new_eilte.shift_start_hour"))
      .withColumn("shift_end_hour", (F.col("new_eilte.shift_end_hour")))
).drop(F.col("new_eilte"))


# Step 1: apply UDF, extract fields, then CACHE + materialize 

#df_udf = df_hours.withColumn("subcategory_name_list", get_category_name(F.col("hours")))


#df_udf.show(truncate=False)
df_elite.show(truncate=False,n=1000)  # Materialization happens at this stage

# Step 2: explode AFTER the UDF stage is fully materialized

```
