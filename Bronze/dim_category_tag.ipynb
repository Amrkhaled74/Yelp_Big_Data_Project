```python


def get_category_name(text):

    if text is None:
       return None, []
    
    list_of_element = text.split(",")
    category_name=list_of_element[0].strip()
    del list_of_element[0]

    if len(list_of_element) == 0:
        return category_name, ["Not Specified"]

    list_of_element_without_space=[st.strip() for st in list_of_element]

    return category_name,list_of_element_without_space  

schema = StructType([
    StructField("Categorie_name", StringType(), True),
    StructField("last_elite_year", ArrayType(StringType()), True),  # <-- fixed
])

get_category_name_udf = F.udf(get_category_name, schema)

df_elite = (
    df_read_business_data.withColumn("new_eilte", get_category_name_udf(F.col("categories")))
      .withColumn("Categorie_name", F.col("new_eilte.Categorie_name"))
      .withColumn("last_elite_year", F.explode(F.col("new_eilte.last_elite_year")))
).drop(F.col("new_eilte"))


df_elite.select(F.col("Categorie_name"), F.col("last_elite_year")).show(n=100, truncate=False)
```
